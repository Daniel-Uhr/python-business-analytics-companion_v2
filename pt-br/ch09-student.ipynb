## Contexto — Rede de farmácias

A rede fictícia de farmácias tem 60 lojas em cada uma de quatro regiões (Sul, Sudeste, Nordeste e Norte). Para avaliar um cupom de desconto, ela **sorteou**, dentro de cada região, 30 lojas para oferecê-lo durante um mês. A tabela registra a região, a área de vendas (m²), se a loja ofereceu o cupom (1) ou não (0) e as vendas do mês (R\$ mil).

O cupom custa à rede cerca de **R\$ 4 mil por loja** no mês e a gestão fixou uma meta de vendas adicionais desse mesmo valor. Os dados são didáticos.

Execute a célula abaixo antes de começar.

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

rng = np.random.default_rng(2026)
regioes = ["Sul", "Sudeste", "Nordeste", "Norte"]
drogarias = pd.DataFrame({"regiao": np.repeat(regioes, 60)})
drogarias["area_m2"] = np.round(rng.uniform(150, 400, size=240))
drogarias["cupom"] = np.concatenate([rng.permutation(np.repeat([0, 1], 30)) for _ in regioes])
nivel_regiao = drogarias["regiao"].map({"Sul": 80, "Sudeste": 95, "Nordeste": 70, "Norte": 65})
efeito_cupom = drogarias["regiao"].map({"Sul": 4, "Sudeste": 6, "Nordeste": 10, "Norte": 2})
drogarias["vendas"] = np.round(nivel_regiao + 0.1 * drogarias["area_m2"] + efeito_cupom * drogarias["cupom"]
                               + rng.normal(0, 6, size=240), 1)
```
