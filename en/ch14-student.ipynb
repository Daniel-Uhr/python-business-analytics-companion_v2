The marketplace observes 400 sellers during one month. Exposure is orders in hundreds. Badge status reflects seller history, and fashion sellers may have higher returns.

```python
import numpy as np
import pandas as pd
pd.set_option("display.width", 160)
pd.set_option("display.max_columns", None)
import statsmodels.formula.api as smf

rng = np.random.default_rng(2026)
n_vendedores = 400
vendedores = pd.DataFrame({
    "badge": rng.binomial(1, 0.4, size=n_vendedores),
    "fashion": rng.binomial(1, 0.3, size=n_vendedores),
    "orders_hundreds": np.round(rng.uniform(0.5, 5, size=n_vendedores), 1),
})
heterogeneage = rng.gamma(0.8, 1 / 0.8, size=n_vendedores)
taxa_lab = np.exp(np.log(2) - 0.3 * vendedores["badge"] + 0.6 * vendedores["fashion"])
vendedores["returns_count"] = rng.poisson(vendedores["orders_hundreds"] * taxa_lab * heterogeneage)
```
