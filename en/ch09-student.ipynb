## The job platform

Study monthly salaries for 400 vacancies. Use the posting as the unit of observation.

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

rng = np.random.default_rng(2026)
n_vagas = 400
vagas = pd.DataFrame({"sector": rng.choice(["retail", "industry", "services", "technology"], size=n_vagas, p=[0.3, 0.25, 0.3, 0.15])})
vagas["experience_years"] = np.round(rng.uniform(0, 10, size=n_vagas), 1)
vagas["remote"] = (rng.random(n_vagas) < 0.25).astype(int)
adicional_sector = vagas["sector"].map({"retail": 0, "industry": 400, "services": 200, "technology": 2_500})
vagas["salary"] = np.round(2_000 + adicional_sector + 250 * vagas["experience_years"] + 300 * vagas["remote"]
                            + rng.normal(0, 500, size=n_vagas), -1)
```

## The shoe retailer

Use observations from 300 stores to assess digital advertising. With a 40% contribution margin, one advertising dollar requires \$2.50 of incremental sales to break even.

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

rng = np.random.default_rng(2026)
n_stores = 300
stores = pd.DataFrame({"mall": (rng.random(n_stores) < 0.4).astype(int)})
stores["digital"] = np.round(rng.uniform(2, 20, size=n_stores), 2)
stores["income_thousands"] = np.round(rng.uniform(1.5, 6, size=n_stores), 2)
stores["sales"] = np.round(100 + 20 * stores["mall"] + 1.0 * stores["digital"] + 1.5 * stores["digital"] * stores["mall"]
                           + 8 * stores["income_thousands"] + 0.5 * stores["digital"] * stores["income_thousands"] + rng.normal(0, 15, size=n_stores), 1)
```

## The pharmacy chain

A one-month experiment assigned half of the 60 stores in each region to offer a coupon. Assess monthly sales in thousands of dollars, controlling for store area.

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

rng = np.random.default_rng(2026)
regions = ["West", "South", "Midwest", "Northeast"]
drogarias = pd.DataFrame({"region": np.repeat(regions, 60)})
drogarias["area_m2"] = np.round(rng.uniform(150, 400, size=240))
drogarias["coupon"] = np.concatenate([rng.permutation(np.repeat([0, 1], 30)) for _ in regions])
nivel_region = drogarias["region"].map({"West": 80, "South": 95, "Midwest": 70, "Northeast": 65})
effect_coupon = drogarias["region"].map({"West": 4, "South": 6, "Midwest": 10, "Northeast": 2})
drogarias["sales"] = np.round(nivel_region + 0.1 * drogarias["area_m2"] + effect_coupon * drogarias["coupon"]
                               + rng.normal(0, 6, size=240), 1)
```
