## The fuel-station chain

For 80 stations the chain records daily liters sold, own price per liter, the nearest competitor’s price and traffic in thousands of vehicles per day. The laboratory has homoskedastic disturbances. Run the setup before analyzing the stations.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf
from scipy import stats

rng = np.random.default_rng(2026)
n_postos = 80
postos = pd.DataFrame({"price": np.round(rng.uniform(5.5, 6.5, size=n_postos), 2)})
postos["competitor_price"] = np.round(postos["price"] + rng.normal(0, 0.15, size=n_postos), 2)
postos["traffic_thousands"] = np.round(rng.uniform(5, 25, size=n_postos), 1)
postos["liters"] = np.round(
    9_000 - 2_500 * postos["price"] + 1_800 * postos["competitor_price"] + 300 * postos["traffic_thousands"]
    + rng.normal(0, 600, size=n_postos)
)
```

## The contact center

The center records in-person and online training hours during the semester, years of experience and cases resolved per day for 150 agents. Agents who take more hours in one format often take more in the other. The simulated data have constant error variance.

```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf
from scipy import stats

rng = np.random.default_rng(2026)
n_atendentes = 150
equipe = pd.DataFrame({"experience_years": np.round(rng.uniform(0, 10, size=n_atendentes), 1)})
equipe["in_person_h"] = np.round(np.clip(25 - 2 * equipe["experience_years"] + rng.normal(0, 4, size=n_atendentes), 0, None), 1)
equipe["online_h"] = np.round(np.clip(0.6 * equipe["in_person_h"] + rng.normal(0, 1.5, size=n_atendentes), 0, None), 1)
equipe["productivity"] = np.round(
    10 + 0.12 * equipe["in_person_h"] + 0.12 * equipe["online_h"] + 1.2 * equipe["experience_years"]
    + rng.normal(0, 2, size=n_atendentes), 1
)
```

## Parrot Cellars

For 200 wines the retailer randomly changes bottle prices by −10, zero or +10 dollars for one month. Quality and baseline prices are recorded before assignment. Management will raise prices only if a ten-dollar increase costs fewer than 50 monthly sales per wine.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.formula.api as smf

rng = np.random.default_rng(2026)
n_rotulos = 200
vinhos = pd.DataFrame({"quality": np.round(rng.uniform(70, 95, size=n_rotulos), 1)})
vinhos["base_price"] = np.round(np.clip(-240 + 4 * vinhos["quality"] + rng.normal(0, 12, size=n_rotulos), 20, None), 2)
vinhos["change"] = rng.choice([-10, 0, 10], size=n_rotulos)
vinhos["price"] = vinhos["base_price"] + vinhos["change"]
vinhos["sales"] = np.round(-1_900 - 6 * vinhos["price"] + 35 * vinhos["quality"] + rng.normal(0, 40, size=n_rotulos)).astype(int)
```

Interpretation conditions: one assignment per wine, complete outcomes, implementation of the assigned price, and no spillovers across wines. Random assignment balances prior characteristics over repetitions, not exactly in each sample. Controls are selected before inspecting outcomes; the linear-response specification and conventional standard errors still require their own assumptions. The 50-bottle threshold is supplied, not estimated from costs.
